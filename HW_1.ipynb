```python
'''
Чтобы узнать, является ли год високосным или нет, можно воспользоваться следующим алгоритмом:
Определить год, который нужно проверить. Можно рассмотреть прошлый, текущий или будущий год.
* Разделить год на 4.
    Если деление целочисленное, то есть без остатка, год високосный.
    Если при делении получился остаток (или десятичная дробь), год не високосный.
* Выяснить, делится ли год на 100.
    Если год делится нацело на 4, но не делится нацело на 100, это високосный год.
    Если год делится как на 4, так и на 100, возможно, это не високосный год, поэтому придётся выполнить ещё одно вычисление.
* Проверить, делится ли год на 400.
    Если год делится на 100 и не делится на 400, это не високосный год.
    Если год делится на 100 и на 400, это високосный год.

    year % 100 - не высокосный
    year % 400 - высокосный
    year % 4 -

'''
```

```python
year = int(input("Введите год: "))
if (year % 100 == 0 and year % 400 != 0) or year % 4 == 0:
  print("Высокосный - ", year)
else:
  print("Не высокосный - ", year)
```
